```python
import pandas as pd

# Step 5: Create your dataframe
df = pd.DataFrame({
    "Character": ["Aragorn", "Boromir", "Frodo", "Galadriel"],
    "Base_Age": [88, 41, 51, 8382]
})

# Step 6: Lambda Mathematical Procedure (e.g., divide age by 10)
df['Valinor_Age'] = df['Base_Age'].apply(lambda x: x / 10)

# Step 7: Lambda Flag Column (e.g., label as 'Hobbit' or 'Other')
df['Race_Flag'] = df['Character'].apply(lambda x: 'Hobbit' if x == 'Frodo' else 'Other')

# Display the result
df
```
